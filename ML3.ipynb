Implement **Gradient Descent Algorithm** to find the local minima of a function. For example, find the
local minima of the function y=(x+3)² starting from the point x=2
Ref: https://builtin.com/data-science/gradient-descent

```python

import numpy as np
import matplotlib.pyplot as plt

# Define function and its derivative
def function(x):
    return (x + 3)**2

def derivative(x):
    return 2 * (x + 3)
```

```python
# Initialize parameters
x = 10
learning_rate = 0.1
iterations = 50

# Gradient Descent
for i in range(iterations):
    x = x - learning_rate * derivative(x)

print("Local Minimum x:", x)
print("Minimum Value y:", function(x))
```

```python
# Plot function
x_values = np.linspace(-6, 4, 100)
y_values = function(x_values)

plt.plot(x_values, y_values, label="y = (x+3)^2")
plt.scatter(x, function(x), color="red", label="Local Minimum")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent")
plt.legend()
plt.grid()
plt.show()
```
